```python
home_wins = (df["FTR"] == "H").sum()
draws = (df["FTR"] == "D").sum()
away_wins = (df["FTR"] == "A").sum()

total = len(df)

print(f"Heimsiege: {home_wins} ({home_wins / total * 100:.2f}%)")
print(f"Unentschieden: {draws} ({draws / total * 100:.2f}%)")
print(f"Auswärtssiege: {away_wins} ({away_wins / total * 100:.2f}%)")
```
